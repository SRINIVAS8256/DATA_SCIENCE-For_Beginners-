```python
v1=[-2,-3,-4,-1]
v2=[2,3,4,1]
cost=0
for i in range(0,len(v1)):
    cost=cost+(v1[i] * v2[i])
print(cost)
```

```python
import numpy as np
```

```python
print(np.dot(v1,v2))
```

```python
# cosine similarity
similarity = (np.dot(v1,v2))/(np.linalg.norm(v1) * np.linalg.norm(v2) )
print(similarity)
```
