# Simple linear regression

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('C:/Users/Admin/OneDrive/Desktop/workspace/DataScience/DS_raw_work/DataSets/placement.csv')
```

```python
df.head(3)
```

```python
plt.scatter(df['cgpa'] , df['package'])
```

```python
from sklearn.model_selection import train_test_split
```

```python
x=df.iloc[:,0:1]
y=df.iloc[:,-1]
```

```python
print(y)
```

# Train Test splitting

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
# print(x_train)
# print(x_test)
# print(y_train)
# print(y_test)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
```

# applying the model

```python
lr.fit(x_train,y_train)
```

# predicting the result

```python
lr.predict(x_test)
```

```python
print(x_test.iloc[0])
```

# adding the predicted values to new array

```python
new_y=np.array([])
for i in range(len(x_test)):
    predict_v = lr.predict(x_test.iloc[i].values.reshape(1,1))[0]
    new_y = np.append(new_y,predict_v)
    true_v = y_test.iloc[i]
    
    print(f"predict_v = {predict_v} and true_v = {true_v} error = {true_v - predict_v}")
```

# Relating wrt straight line y=mx + b    m=slope   b = intercept

```python
m=lr.coef_
b=lr.intercept_
print(f"m = {m} and b = {b}")
```
